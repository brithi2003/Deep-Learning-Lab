```python
import torch
import torch.nn as nn

class TCN(nn.Module):

    def __init__(self):

        super().__init__()

        self.conv1 = nn.Conv1d(
            1, 16,
            kernel_size=3,
            padding=2,
            dilation=1
        )

        self.conv2 = nn.Conv1d(
            16, 32,
            kernel_size=3,
            padding=4,
            dilation=2
        )

        self.conv3 = nn.Conv1d(
            32, 16,
            kernel_size=3,
            padding=8,
            dilation=4
        )

        self.fc = nn.Linear(16, 1)

    def forward(self, x):

        x = torch.relu(
            self.conv1(x)
        )

        x = torch.relu(
            self.conv2(x)
        )

        x = torch.relu(
            self.conv3(x)
        )

        x = x[:, :, -1]

        return self.fc(x)


model = TCN()

x = torch.randn(
    8, 1, 50
)

output = model(x)

print("Input shape :", x.shape)
print("Output shape:", output.shape)
```
