```python
import torch

torch.manual_seed(42)

seq_len = 5
input_size = 3
hidden_size = 4

X = torch.randn(seq_len, input_size)

W = torch.randn(input_size + hidden_size, hidden_size * 4)
b = torch.zeros(hidden_size * 4)

def lstm_forward(X):

    h = torch.zeros(hidden_size)
    c = torch.zeros(hidden_size)

    outputs = []

    for x in X:

        combined = torch.cat([h, x])

        gates = combined @ W + b

        i, f, g, o = torch.chunk(gates, 4)

        i = torch.sigmoid(i)
        f = torch.sigmoid(f)
        g = torch.tanh(g)
        o = torch.sigmoid(o)

        c = f*c + i*g
        h = o*torch.tanh(c)

        outputs.append(h)

    return torch.stack(outputs)


forward = lstm_forward(X)
backward = torch.flip(
    lstm_forward(torch.flip(X, [0])),
    [0]
)

bilstm_output = torch.cat(
    [forward, backward],
    dim=1
)

print("BiLSTM output:")
print(bilstm_output)
print("Output shape:", bilstm_output.shape)
```
