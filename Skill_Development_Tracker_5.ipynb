```python
import torch

X = torch.randn(4, 5)

# Batch Normalization
batch_mean = X.mean(dim=0, keepdim=True)
batch_var = X.var(dim=0, unbiased=False, keepdim=True)

batch_norm = (
    X - batch_mean
) / torch.sqrt(batch_var + 1e-5)

# Layer Normalization
layer_mean = X.mean(dim=1, keepdim=True)
layer_var = X.var(dim=1, unbiased=False, keepdim=True)

layer_norm = (
    X - layer_mean
) / torch.sqrt(layer_var + 1e-5)

print("Input:")
print(X)

print("\nBatch Normalization:")
print(batch_norm)

print("\nLayer Normalization:")
print(layer_norm)
```
