```python
import torch
import torch.distributed as dist
import torch.nn as nn

def main():

    dist.init_process_group(
        backend="gloo",
        rank=0,  # Explicitly set rank for single-process execution
        world_size=1,  # Explicitly set world_size for single-process execution
        init_method="tcp://localhost:12355" # Specify an init_method for local execution
    )

    rank = dist.get_rank()

    model = nn.Linear(10, 1)

    x = torch.randn(5, 10)
    y = model(x)

    print(
        "Process:",
        rank,
        "Output shape:",
        y.shape
    )

    dist.destroy_process_group()


if __name__ == "__main__":
    main()
```
