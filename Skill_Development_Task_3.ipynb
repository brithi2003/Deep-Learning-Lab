```python
import numpy as np

def im2col(X, kernel_size, stride=1, padding=0):

    X = np.pad(
        X,
        ((0,0),(0,0),
         (padding,padding),
         (padding,padding))
    )

    N, C, H, W = X.shape
    K = kernel_size

    out_h = (H-K)//stride + 1
    out_w = (W-K)//stride + 1

    cols = []

    for i in range(0, H-K+1, stride):
        for j in range(0, W-K+1, stride):
            patch = X[:, :, i:i+K, j:j+K]
            cols.append(patch.reshape(N, -1))

    return np.stack(cols, axis=1)


# Input image
X = np.random.randn(1, 3, 5, 5)

# Filter
W = np.random.randn(2, 3, 3, 3)

cols = im2col(X, 3)

W_col = W.reshape(2, -1)

output = cols @ W_col.T

output = output.transpose(0, 2, 1)

print("Input shape :", X.shape)
print("Output shape:", output.shape)
print(output)
```
