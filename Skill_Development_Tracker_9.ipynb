```python
import torch

def ciou_loss(box1, box2):

    # box = x1, y1, x2, y2

    x1 = torch.max(box1[0], box2[0])
    y1 = torch.max(box1[1], box2[1])

    x2 = torch.min(box1[2], box2[2])
    y2 = torch.min(box1[3], box2[3])

    intersection = torch.clamp(
        x2-x1, min=0
    ) * torch.clamp(
        y2-y1, min=0
    )

    area1 = (
        box1[2]-box1[0]
    ) * (
        box1[3]-box1[1]
    )

    area2 = (
        box2[2]-box2[0]
    ) * (
        box2[3]-box2[1]
    )

    union = area1 + area2 - intersection

    iou = intersection / (union + 1e-7)

    return 1 - iou


pred_box = torch.tensor(
    [10., 10., 50., 50.]
)

true_box = torch.tensor(
    [15., 15., 55., 55.]
)

loss = ciou_loss(
    pred_box,
    true_box
)

print("Box Regression Loss:", loss.item())
```
