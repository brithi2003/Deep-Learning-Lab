```python
import numpy as np
import matplotlib.pyplot as plt

# Swish activation function
def swish(x):
    return x / (1 + np.exp(-x))


# Create input values
x = np.linspace(-5, 5, 100)

# Calculate Swish
y = swish(x)

# Print some values
print("Input:")
print(x[:10])

print("\nSwish Output:")
print(y[:10])


# Plot Swish function
plt.plot(x, y)

plt.xlabel("Input")
plt.ylabel("Swish Output")
plt.title("Swish Activation Function")

plt.grid(True)
plt.show()
```
