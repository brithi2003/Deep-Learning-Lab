```python
import torch
import torch.nn as nn

class Autoencoder(nn.Module):

    def __init__(self):

        super().__init__()

        self.encoder = nn.Sequential(
            nn.Conv2d(1, 16, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(16, 8, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2)
        )

        self.decoder = nn.Sequential(
            nn.ConvTranspose2d(
                8, 16, 2, stride=2
            ),
            nn.ReLU(),

            nn.ConvTranspose2d(
                16, 1, 2, stride=2
            ),
            nn.Sigmoid()
        )

    def forward(self, x):

        encoded = self.encoder(x)
        decoded = self.decoder(encoded)

        return decoded


model = Autoencoder()

x = torch.rand(
    2, 1, 28, 28
)

output = model(x)

loss = torch.mean(
    (output-x)**2
)

print("Input :", x.shape)
print("Output:", output.shape)
print("Reconstruction error:", loss.item())
```
